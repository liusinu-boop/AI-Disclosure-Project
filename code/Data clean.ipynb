```python
import pandas as pd
import statsmodels.api as sm
import numpy as np

# 1. 读取数据
file_path = "/Users/snowiiy/Desktop/26 spring/MGS3001/假设1数据/H1_final_dataset_correct_Y.csv"
df = pd.read_csv(file_path)

# 2. 删除 AI disclosure = 0 的公司
df = df[df['ai_disclosure_sentence_ratio'] > 0]

# 3. 替换 inf 为 NaN
df = df.replace([np.inf, -np.inf], np.nan)

# 4. 删除含 NaN 的行
df = df.dropna()

# 5. 设置自变量 X 和因变量 Y
Y = df['turnover']
X_vars = ['ai_disclosure_sentence_ratio', 'Size', 'Lev', 'ROA', 'Growth', 
          'CashRatio', 'TobinQ', 'ListAge', 'Board', 'Indep', 'Dual', 'Top1', 'SOE']
X = df[X_vars]

# 6. 加入常数项
X = sm.add_constant(X)

# 7. 回归
model = sm.OLS(Y, X).fit()

# 8. 输出结果
print(model.summary())
```
